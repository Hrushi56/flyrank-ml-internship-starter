# 1. My Lane and Why

I selected the **Content Refresh** lane.

Many websites contain thousands of pages, making it difficult for content teams to manually decide which pages should be updated first. The dataset contains information about content age, traffic trends, search performance, and engagement metrics. These features make it possible to identify pages that are likely to benefit from a refresh.

This project focuses on helping teams prioritize their content update efforts instead of relying on manual review.

# 2. The Research Question

## Research Question

Which content pages should be prioritized for refresh to maximize future search performance and user engagement?

## Decision

Decide which pages should be refreshed first.

## Action

The content team updates the highest-priority pages before spending time on lower-impact pages.

## Cost of a Wrong Recommendation

If pages are incorrectly prioritized, the team may spend time updating pages that provide little benefit while important pages continue losing visibility and traffic.

## Why Machine Learning Can Help

With thousands of pages and many performance metrics, manually identifying the best candidates for refresh is difficult. Machine learning can identify patterns across multiple variables to support better prioritization.

## 3. Quick Look at the Data

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print("Total Content Pages:", len(df))

print("\nTrend Direction:")
print(df["trend_direction"].value_counts())

print("\nFreshness Tier:")
print(df["freshness_tier"].value_counts())

print("\nAverage CTR:", round(df["ctr"].mean(), 2))
print("Average Position:", round(df["avg_position"].mean(), 2))
```

### Interpretation

The dataset contains **30,000 content pages**, which is large enough that manual review is impractical.

More than **16,000 pages are trending down**, suggesting many pages may benefit from updates.

The average CTR is approximately **0.51**, while the average search position is about **16.34**, indicating there is room to improve visibility through better content prioritization.

## 4. Careful Words

This project will recommend which pages should be reviewed for refresh based on historical data.

It does not guarantee that refreshing a page will increase traffic, since search rankings also depend on factors outside this dataset such as competition, search intent, and search engine algorithm updates.

The recommendations should support human decision-making rather than replace it.
