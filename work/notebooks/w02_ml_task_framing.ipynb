```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print("Dataset Shape:", df.shape)
```

# 1. My Lane as an ML Task

## Selected Lane

Content Refresh

## ML Task Type

Ranking (or Scoring)

## Why?

The goal is not simply to classify pages as "refresh" or "don't refresh." Instead, the objective is to rank content pages according to their priority for updating. Pages with higher priority scores should be reviewed first by the content team.

A ranking approach is more useful because resources are limited and not every page can be refreshed at the same time.

```python
# ML Task Type

lane = "Content Refresh"
ml_task = "Ranking"

print("Selected Lane:", lane)
print("ML Task Type:", ml_task)
```

# 2. Target or Proxy

The dataset does not contain a direct column indicating whether a page should be refreshed.

Instead, a proxy target can be created using existing performance indicators such as:

- trend_direction
- trend_pct
- impressions_last_30d
- impressions_prev_30d
- clicks_last_30d
- ctr
- avg_position

Pages with declining traffic, worsening rankings, or decreasing engagement are more likely to require a refresh.

The proxy target represents the refresh priority rather than an absolute ground truth.

```python
# Looking at possible target/proxy columns

target_columns = [
    "trend_direction",
    "trend_pct",
    "impressions_last_30d",
    "impressions_prev_30d",
    "clicks_last_30d",
    "ctr",
    "avg_position"
]

df[target_columns].head()
```

# 3. Success Metric

The success of the model will be evaluated by how accurately it ranks pages that should be refreshed first.

Possible evaluation metrics include:

- Precision@K
- Recall@K
- Mean Average Precision (MAP)

From a business perspective, success means that the highest-ranked pages provide measurable improvements after being refreshed.

```python
print("Possible Evaluation Metrics")

metrics = [
    "Precision@K",
    "Recall@K",
    "Mean Average Precision (MAP)"
]

for metric in metrics:
    print("-", metric)
```

# 4. Unit of Analysis

Each row in the dataset represents one content page.

The model makes one recommendation for each page.

Example attributes include:

- Content ID
- Search volume
- Traffic trend
- Click-through rate
- Average search position

- Freshness tier

These features describe the current condition of a page and help estimate whether it should be refreshed.

```python
unit_columns = [
    "content_id",
    "search_volume",
    "trend_direction",
    "trend_pct",
    "ctr",
    "avg_position",
    "content_age_days",
    "freshness_tier"
]

unit_df = df[unit_columns]

print("Each row represents one content page.\n")

unit_df.head()
```

# 5. Why ML Beats a Fixed Rule

A simple rule such as:

"If traffic decreased by 20%, refresh the page"

would ignore many important factors.

For example:

- Search competition
- Page age
- Search position
- User engagement
- Search volume
- Historical performance

Machine learning can combine all of these variables to identify patterns that are difficult to capture with manually designed rules.

Therefore, ML provides more flexible and data-driven recommendations than a fixed threshold.

```python
features = [
    "search_volume",
    "ctr",
    "avg_position",
    "content_age_days",
    "trend_pct",
    "engagement_rate"
]

print("Features that ML can combine:")

for feature in features:
    print("-", feature)
```
